**Objective**:
Analyze how product pricing, customer purchase patterns, and seasonal trends impact sales in an online retail business.

Business use:

* Identify top-performing products
* Understand price sensitivity
* Find revenue-driving customer segments

```python
import pandas as pd

filePath = '/content/online_retail_II.xlsx'
xls = pd.read_excel(filePath, sheet_name=None)
df = pd.concat(xls.values(), ignore_index=True)
df.head()
```

Basic **Cleaning**

```python
# Standardize column names
df.columns = [c.strip() for c in df.columns]
df.columns = [c.replace(' ', '_').replace('.', '') for c in df.columns]

# Convert InvoiceDate to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')
```

```python
# Remove rows with missing critical info
df = df[~df['InvoiceDate'].isna()]
df = df.dropna(subset=['Price', 'Quantity'])

# Remove transactions with UnitPrice <= 0 (bad data) and Quantity == 0
df = df[(df['Price'] > 0)]
```

```python
# Separate returns for analysis (Quantity < 0)
returns = df[df['Quantity'] < 0].copy()
sales = df[df['Quantity'] > 0].copy()
print("Sales rows:", sales.shape[0], "Returns rows:", returns.shape[0])
```

**Feature Engineering**

```python
sales['Revenue'] = sales['Quantity'] * sales['UnitPrice']
sales['InvoiceMonth'] = sales['InvoiceDate'].dt.to_period('M').astype(str)
sales['InvoiceDay'] = sales['InvoiceDate'].dt.date
sales['InvoiceHour'] = sales['InvoiceDate'].dt.hour
sales['DayOfWeek'] = sales['InvoiceDate'].dt.day_name()
sales['ItemKey'] = sales['StockCode'].astype(str)

sales.head(10)
```

```python
# RFM features (customer-level)
cust = sales.dropna(subset=['Customer_ID'])

# Convert customerId to int
cust['Customer_ID'] = cust['Customer_ID'].astype(int)

cust['Revenue'] = cust['Quantity'] * cust['Price']
snapshot_date = cust['InvoiceDate'].max() + pd.Timedelta(days=1)
rfm = cust.groupby('Customer_ID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'Invoice': 'nunique',
    'Revenue': 'sum'
}).reset_index()
rfm.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

rfm.head()
```

**Exploratory Data Analysis (Plots)**

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Top 20 products by revenue
product_stats = sales.groupby('ItemKey').agg(
    total_units_sold=('Quantity', 'sum'),
    total_revenue=('Revenue', 'sum'),
    avg_unit_price=('Price', 'mean'),
    transactions=('Invoice', 'nunique')
).reset_index().sort_values('total_revenue', ascending=False)

# Get unique product descriptions (StockCode → Description)
product_desc = sales.groupby('ItemKey')['Description'].first().reset_index()

# Merge into product_stats
product_stats = product_stats.merge(product_desc, on='ItemKey', how='left')

product_stats['ShortDesc'] = product_stats['Description'].str.slice(0, 60) + "..."


top20 = product_stats.head(20).copy()
plt.figure(figsize=(10,6))

sns.barplot(data=top20, x='total_revenue', y='ShortDesc', palette='Blues_r')

plt.title('Top 20 Products by Revenue', fontsize=14)
plt.xlabel('Total Revenue')
plt.ylabel('Product Description')

plt.tight_layout()
plt.show()
```

```python

# Revenue by month
monthly = sales.groupby('InvoiceMonth').agg(month_revenue=('Revenue','sum')).reset_index()
monthly['InvoiceMonth'] = pd.to_datetime(monthly['InvoiceMonth'])
monthly = monthly.sort_values('InvoiceMonth')
plt.figure(figsize=(12,5))
sns.lineplot(data=monthly, x='InvoiceMonth', y='month_revenue', marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Price distribution
plt.figure(figsize=(8,5))
sns.histplot(sales['Price'].clip(upper=sales['Price'].quantile(0.99)), bins=50, kde=False)
plt.title('Unit Price Distribution (capped at 99th percentile)')
plt.xlabel('UnitPrice')
plt.tight_layout()
plt.show()
```

```python
# Scatter: avg price vs total units sold (per product)
agg = product_stats.copy()
plt.figure(figsize=(8,6))
sns.scatterplot(data=agg, x='avg_unit_price', y='total_units_sold', size='total_revenue', alpha=0.7)
plt.xscale('log')  # skew correction
plt.yscale('log')
plt.xlabel('Avg Unit Price (log)')
plt.ylabel('Total Units Sold (log)')
plt.title('Price vs Units Sold (product-level, log-log)')
plt.tight_layout()
plt.show()
```

```python
# Revenue by Country (top 10)
country_rev = sales.groupby('Country').agg(total_revenue=('Revenue','sum')).reset_index()
top_countries = country_rev.sort_values('total_revenue', ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(data=top_countries, x='total_revenue', y='Country')
plt.title('Top 10 Countries by Revenue')
plt.xlabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
# RFM: distribution plots
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
sns.histplot(rfm['Recency'], bins=50)
plt.title('Recency (days)')
plt.subplot(1,3,2)
sns.histplot(rfm['Frequency'], bins=50)
plt.title('Frequency (# invoices)')
plt.subplot(1,3,3)
sns.histplot(rfm['Monetary'], bins=50)
plt.title('Monetary (total revenue)')
plt.tight_layout()
plt.show()
```
